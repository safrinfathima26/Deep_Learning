```python
import torch
import torch.nn as nn

# Generator
# -----------------------------
generator = nn.Sequential(
    nn.Linear(10, 20),
    nn.ReLU(),
    nn.Linear(20, 5)
)

# Critic
critic = nn.Sequential(
    nn.Linear(5, 20),
    nn.ReLU(),
    nn.Linear(20, 1)
)

# Real Data
real_data = torch.randn(1, 5)

# Random Noise
noise = torch.randn(1, 10)

# Generate Fake Data
fake_data = generator(noise)

# -----------------------------
# Critic Scores
# -----------------------------
real_score = critic(real_data)
fake_score = critic(fake_data)

# Wasserstein Loss
critic_loss = fake_score.mean() - real_score.mean()

generator_loss = -fake_score.mean()

# Simple Gradient Penalty
gradient_penalty = torch.mean((real_score - fake_score) ** 2)

# Total Critic Loss
total_critic_loss = critic_loss + 10 * gradient_penalty

# Results
print("Generated Fake Data")
print(fake_data)

print("\nReal Score")
print(real_score)

print("\nFake Score")
print(fake_score)

print("\nGenerator Loss")
print(generator_loss.item())

print("\nCritic Loss")
print(critic_loss.item())

print("\nGradient Penalty")
print(gradient_penalty.item())

print("\nTotal Critic Loss")
print(total_critic_loss.item())
```
