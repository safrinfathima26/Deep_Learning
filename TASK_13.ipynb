```python
import torch
import torch.nn as nn

# Input Sequence
X = torch.randn(1, 1, 10)

# Simple TCN Block
class TCNBlock(nn.Module):

    def __init__(self):

        super().__init__()

        self.conv = nn.Conv1d(
            in_channels=1,
            out_channels=1,
            kernel_size=3,
            dilation=2,
            padding=4       # Causal padding
        )

        self.relu = nn.ReLU()

    def forward(self, x):

        # Save input for residual connection
        residual = x

        # Dilated Convolution
        out = self.conv(x)

        # Remove extra values added by padding
        out = out[:, :, :x.size(2)]

        # Residual Connection
        out = out + residual

        # Activation
        out = self.relu(out)

        return out

# Create Model
tcn = TCNBlock()

# Forward Pass
features = tcn(X)

# Sequence Prediction Layer
predictor = nn.Conv1d(
    in_channels=1,
    out_channels=1,
    kernel_size=1
)

prediction = predictor(features)

# Results
print("Input Shape      :", X.shape)
print("Feature Shape    :", features.shape)
print("Prediction Shape :", prediction.shape)

print("\nPrediction Output:")
print(prediction)
```
