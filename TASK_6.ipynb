```python
import torch

# Input Sequence
X = torch.tensor([1.0, 2.0, 3.0, 4.0])

# Initial Hidden State
h = torch.tensor(0.0)

# Initial Cell State
c = torch.tensor(0.0)


# LSTM Cell
def lstm_cell(x, h, c):

    # Input Gate
    i = torch.sigmoid(x + h)

    # Forget Gate
    f = torch.sigmoid(x + h)

    # Output Gate
    o = torch.sigmoid(x + h)

    # Cell Candidate
    g = torch.tanh(x + h)

    # Cell State
    c = f * c + i * g

    # Hidden State
    h = o * torch.tanh(c)

    return h, c

# Forward Direction
forward_output = []

h = torch.tensor(0.0)
c = torch.tensor(0.0)

for x in X:

    h, c = lstm_cell(x, h, c)

    forward_output.append(h.item())

# Backward Direction
backward_output = []

h = torch.tensor(0.0)
c = torch.tensor(0.0)

for x in reversed(X):

    h, c = lstm_cell(x, h, c)

    backward_output.insert(0, h.item())


# BiLSTM Output
print("Forward Output")
print(forward_output)

print("\nBackward Output")
print(backward_output)

print("\nBiLSTM Output")

for f, b in zip(forward_output, backward_output):

    print([round(f,3), round(b,3)])
```
