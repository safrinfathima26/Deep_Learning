```python
#importing numpy library
import numpy as np

# Same random values every time
np.random.seed(42)

# Input and Output
# Input (4 samples, 2 features)
X = np.array([
    [2, 1],
    [1, 3],
    [4, 5],
    [6, 2]
])

# Output (One-Hot Encoded)
Y = np.array([
    [1, 0],
    [0, 1],
    [1, 0],
    [0, 1]
])

# Initializing Weights and Biases
W1 = np.random.randn(2, 3)
b1 = np.zeros((1, 3))

W2 = np.random.randn(3, 2)
b2 = np.zeros((1, 2))

learning_rate = 0.01
epochs = 100
m = X.shape[0]


# ReLU Activation
def relu(x):
    return np.maximum(0, x)

# ReLU Derivative
def relu_derivative(x):
    return (x > 0).astype(float)

# Softmax Activation
def softmax(x):
    # Numerical stability
    x = x - np.max(x, axis=1, keepdims=True)
    exp = np.exp(x)
    return exp / np.sum(exp, axis=1, keepdims=True)

# Training
for epoch in range(epochs):

    #Forward Pass
    # Hidden Layer
    Z1 = np.dot(X, W1) + b1
    H = relu(Z1)

    # Output Layer
    Z2 = np.dot(H, W2) + b2
    O = softmax(Z2)

    loss = -np.sum(Y * np.log(O + 1e-8)) / m

    #Backward Pass

    # Output Layer Error
    dZ2 = O - Y

    dW2 = np.dot(H.T, dZ2) / m
    db2 = np.sum(dZ2, axis=0, keepdims=True) / m

    # Hidden Layer Error
    dH = np.dot(dZ2, W2.T)

    dZ1 = dH * relu_derivative(Z1)

    dW1 = np.dot(X.T, dZ1) / m
    db1 = np.sum(dZ1, axis=0, keepdims=True) / m

    #Updating weights

    W2 = W2 - learning_rate * dW2
    b2 = b2 - learning_rate * db2

    W1 = W1 - learning_rate * dW1
    b1 = b1 - learning_rate * db1

    # Printing Loss Every 10 Epochs
    if epoch % 10 == 0:
        print(f"Epoch {epoch:3d}  Loss = {loss:.4f}")

# Final Prediction
print("\nFinal Output (Softmax Probabilities):")
print(O)

print("\nPredicted Class:")
print(np.argmax(O, axis=1))

print("\nActual Class:")
print(np.argmax(Y, axis=1))
```
