```python
import torch

# Predicted Bounding Box
pred_box = torch.tensor([2.0, 2.0, 6.0, 6.0])

# Ground Truth Bounding Box
true_box = torch.tensor([3.0, 3.0, 7.0, 7.0])

# IoU Function
def iou(box1, box2):

    x1 = max(box1[0], box2[0])
    y1 = max(box1[1], box2[1])
    x2 = min(box1[2], box2[2])
    y2 = min(box1[3], box2[3])

    intersection = max(0, x2 - x1) * max(0, y2 - y1)

    area1 = (box1[2]-box1[0]) * (box1[3]-box1[1])
    area2 = (box2[2]-box2[0]) * (box2[3]-box2[1])

    union = area1 + area2 - intersection

    return intersection / union

# IoU Loss
iou_score = iou(pred_box, true_box)
iou_loss = 1 - iou_score

# Bounding Box Regression Loss
# (Mean Squared Error)
bbox_loss = torch.mean((pred_box - true_box) ** 2)

# Classification
prediction = torch.tensor([[0.8, 0.2]])
target = torch.tensor([[1.0, 0.0]])

# Mean Squared Error Classification Loss
cls_loss = torch.mean((prediction - target) ** 2)

# Total Detection Loss
total_loss = iou_loss + bbox_loss + cls_loss


# Results
print("IoU Score               :", iou_score.item())
print("IoU Loss                :", iou_loss.item())

print("\nBounding Box Loss       :", bbox_loss.item())

print("\nClassification Loss     :", cls_loss.item())

print("\nTotal Detection Loss    :", total_loss.item())
```
