```python
import torch
import matplotlib.pyplot as plt

# Swish Activation Function
def swish(x):
    return x * torch.sigmoid(x)

# ReLU Activation Function
def relu(x):
    return torch.relu(x)

# Input Values
x = torch.linspace(-5, 5, 100)


# Forward Pass
swish_output = swish(x)
relu_output = relu(x)


# Print Sample Output
print("Input Values:")
print(x[:10])

print("\nSwish Output:")
print(swish_output[:10])

print("\nReLU Output:")
print(relu_output[:10])

# Plot Activation Functions
plt.plot(x.numpy(), swish_output.numpy(), label="Swish")
plt.plot(x.numpy(), relu_output.numpy(), label="ReLU")

plt.title("Swish vs ReLU Activation")
plt.xlabel("Input")
plt.ylabel("Output")
plt.legend()

plt.show()
```
